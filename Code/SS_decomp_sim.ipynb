```python
import numpy as np
import pandas as pd
from plotnine import *
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

# Set the seed for reproducibility
np.random.seed(42)

%config InlineBackend.figure_format = 'retina' 
```

```python
def generate_data(sigma):
    X = np.linspace(start=1, stop=10, num=10).reshape(-1, 1) # Array w/ 1 column, the -1 means 'infer the # of rows'

    # Define parameters
    beta0 = 4
    beta1 = 3
    
    # Generate epsilon as a 1D array of N(0, sigma^2) RVs
    epsilon = sigma * np.random.randn(10)
    
    # Generate y using SLR model
    y = beta0 + beta1 * X.flatten() + epsilon

    return X, y 
```

```python
def ss_decomp(X, y):

    # Fit a linear regression model
    model = LinearRegression()
    model.fit(X, y)

    # Make predictions
    y_pred = model.predict(X)

    # Calculate the mean of y
    y_mean = np.mean(y)
    n = len(y)

    # Calculate SS quantities
    SST = np.sum((y - y_mean) ** 2).round(4)
    SSR = np.sum((y_pred - y_mean) ** 2).round(4)
    SSE = np.sum((y - y_pred) ** 2).round(4)

    F_stat = (SSR/1)/(SSE/(n-2))

    # Output the sum of squares decomposition
    print(f"SST (Total Sum of Squares): {SST}")
    print(f"SSR (Regression Sum of Squares): {SSR}")
    print(f"SSE (Error Sum of Squares): {SSE}")
    print(f"SST = SSR + SSE: {np.isclose(SST, SSR + SSE)}")
    
    print(f"Coefficient of Determination, R^2 is: {np.round(1-SSE/SST, 4)}")
    
    # Make a ggplot
    df = pd.DataFrame({
        'X': X.flatten(),
        'y': y,
        'predicted': y_pred,
        'y_mean':  y_mean
    })

    gg1 = (
        ggplot(df, aes(x = 'X', y = 'y')) + 
        geom_point() + 
        # geom_smooth(method = "lm", formula = "y ~ x", se = False) +
        geom_hline(yintercept=y_mean, linetype='dashed') + 
        geom_segment(aes(xend = 'X', yend = 'y_mean'), color = "black") + # SST components
        ggtitle(f"R^2 is {np.round(1-SSE/SST, 4)}")
        )
    
    gg2 = (
        ggplot(df, aes(x = 'X', y = 'y')) + 
        geom_point() + 
        geom_smooth(method = "lm", formula = "y ~ x", se = False) +
        geom_hline(yintercept=y_mean, linetype='dashed') + 
        geom_segment(aes(xend = 'X', y = 'predicted', yend = 'y_mean'), color = "blue", linetype = "dashed") + # SSR components 
        geom_segment(aes(xend = 'X',  yend = 'predicted'), color = "red", linetype = "dashed") + # SSE components
        ggtitle(f"R^2 is {np.round(1-SSE/SST, 4)}")
        )

    display(gg1, gg2) # Print the plots
    
    return gg1, gg2 # Return the gg objects if you want to store the plots into variables
```

```python
X, y = generate_data(sigma = 2)
ss_decomp(X, y)
```

```python
def generate_data(sigma, beta0, beta1, n):
    X = np.linspace(start = 1, stop = n, num = n).reshape(-1, 1) # Array w/ 1 column, the -1 means 'infer the # of rows'

    # Define parameters
    # beta0 = intercept
    # beta1 = slope
    
    # Generate epsilon as a 1D array of N(0, sigma^2) RVs
    epsilon = sigma * np.random.randn(n)
    
    # Generate y using SLR model
    y = beta0 + beta1 * X.flatten() + epsilon

    return X, y 
```

```python
def ss_decomp(X, y):

    # Fit a linear regression model
    model = LinearRegression()
    model.fit(X, y)

    # Make predictions
    y_pred = model.predict(X)

    # Calculate the mean of y
    y_mean = np.mean(y)
    n = len(y)

    # Calculate SS quantities
    SST = np.sum((y - y_mean) ** 2).round(4)
    SSR = np.sum((y_pred - y_mean) ** 2).round(4)
    SSE = np.sum((y - y_pred) ** 2).round(4)

    # Output the sum of squares decomposition
    print(f"SST (Total Sum of Squares): {SST}")
    print(f"SSR (Regression Sum of Squares): {SSR}")
    print(f"SSE (Error Sum of Squares): {SSE}")
    print(f"SST = SSR + SSE: {np.isclose(SST, SSR + SSE)}")
    
    print(f"Coefficient of Determination, R^2 is: {np.round(1-SSE/SST, 4)}")

    # Calculate F-statistic
    F_stat = (SSR/1)/(SSE/(n-2))

    # Calculate F-value
    from scipy.stats import f
    alpha = 0.05
    df1 = 1
    df2 = n - 2
    F_crit = f.ppf(1 - alpha, df1, df2)

    print(f"The F-Statistic is: {F_stat}")
    print(f"The Critical F-Value (alpha = 0.05) is: {F_crit}")

    if F_stat > F_crit:
        print("We reject the null hypothesis that the slope = 0 . X is a significant predictor of Y")
    else:
        print("We fail to reject the null hypothesis. X is not a significant predictor of Y")
    
    # Make a ggplot
    df = pd.DataFrame({
        'X': X.flatten(),
        'y': y,
        'predicted': y_pred,
        'y_mean':  y_mean
    })

    gg1 = (
        ggplot(df, aes(x = 'X', y = 'y')) + 
        geom_point() + 
        # geom_smooth(method = "lm", formula = "y ~ x", se = False) +
        geom_hline(yintercept=y_mean, linetype='dashed') + 
        geom_segment(aes(xend = 'X', yend = 'y_mean'), color = "black") + # SST components
        ggtitle(f"R^2 is {np.round(1-SSE/SST, 4)}")
        )
    
    gg2 = (
        ggplot(df, aes(x = 'X', y = 'y')) + 
        geom_point() + 
        geom_smooth(method = "lm", formula = "y ~ x", se = False) +
        geom_hline(yintercept=y_mean, linetype='dashed') + 
        geom_segment(aes(xend = 'X', y = 'predicted', yend = 'y_mean'), color = "blue", linetype = "dashed") + # SSR components 
        geom_segment(aes(xend = 'X',  yend = 'predicted'), color = "red", linetype = "dashed") + # SSE components
        ggtitle(f"R^2 is {np.round(1-SSE/SST, 4)}")
        )

    display(gg1, gg2) # Print the plots
    
    return gg1, gg2 # Return the gg objects if you want to store the plots into variables
```

```python
def calculate_f_and_t_statistic(X, y):

    # Fit a linear regression model
    model = LinearRegression()
    model.fit(X, y)

    # Make predictions
    y_pred = model.predict(X)

    # Calculate the mean of y
    y_mean = np.mean(y)
    n = len(y)

    # Calculate SS quantities
    SST = np.sum((y - y_mean) ** 2).round(4)
    SSR = np.sum((y_pred - y_mean) ** 2).round(4)
    SSE = np.sum((y - y_pred) ** 2).round(4)

    # Output the sum of squares decomposition
    print(f"SST (Total Sum of Squares): {SST}")
    print(f"SSR (Regression Sum of Squares): {SSR}")
    print(f"SSE (Error Sum of Squares): {SSE}")
    print(f"SST = SSR + SSE: {np.isclose(SST, SSR + SSE)}")
    
    print(f"Coefficient of Determination, R^2 is: {np.round(1-SSE/SST, 4)}")

    # Calculate F-statistic
    F_stat = (SSR/1)/(SSE/(n-2))

    # Calculate F-value
    from scipy.stats import f
    alpha = 0.05
    df1 = 1
    df2 = n - 2
    F_crit = f.ppf(1 - alpha, df1, df2)

    print(f"The F-Statistic is: {F_stat}")
    print(f"The Critical F-Value (alpha = 0.05) is: {F_crit}")

    if F_stat > F_crit:
        print("We reject the null hypothesis that the slope = 0 . X is a significant predictor of Y")
    else:
        print("We fail to reject the null hypothesis. X is not a significant predictor of Y")
    
        # Calculate t-statistic for slope coefficient
    # Get the slope coefficient
    slope = model.coef_[0]
    
    # Calculate standard error of slope
    MSE = SSE / (n - 2)
    X_centered = X - np.mean(X)
    SE_slope = np.sqrt(MSE / np.sum(X_centered ** 2))
    
    # Calculate t-statistic for testing H0: slope = 0
    t_stat = slope / SE_slope
    
    print(f"The t-Statistic for slope is: {t_stat}")
    
    # Make a ggplot
    df = pd.DataFrame({
        'X': X.flatten(),
        'y': y,
        'predicted': y_pred,
        'y_mean':  y_mean
    })
    
    return t_stat, F_stat
```

```python
# Simulation with B = 100
B = 100
t_stats = []
f_stats = []

for b in range(B):
    # Generate random data
    n = 50  # sample size
    X = np.random.randn(n, 1)
    # Generate y with some relationship to X plus noise
    true_slope = np.random.randn()  # Random true slope
    y = true_slope * X.flatten() + np.random.randn(n)
    
    # Calculate statistics (suppress output for simulation)
    import sys
    from io import StringIO
    
    # Capture output to avoid printing during simulation
    old_stdout = sys.stdout
    sys.stdout = StringIO()
    
    t_stat, f_stat = calculate_f_and_t_statistic(X, y)
    
    # Restore output
    sys.stdout = old_stdout
    
    t_stats.append(t_stat)
    f_stats.append(f_stat)

# Convert to numpy arrays
t_stats = np.array(t_stats)
f_stats = np.array(f_stats)

# Create scatter plot
plt.figure(figsize=(10, 8))
plt.scatter(t_stats, f_stats, alpha=0.7, s=50)
plt.xlabel('t-statistic', fontsize=12)
plt.ylabel('F-statistic', fontsize=12)
plt.title('Relationship between t-statistic and F-statistic\n(B = 100 simulations)', fontsize=14)
plt.grid(True, alpha=0.3)

# Add the theoretical relationship line: F = t^2
t_range = np.linspace(min(t_stats), max(t_stats), 100)
plt.plot(t_range, t_range**2, 'r--', linewidth=2, label='F = t²')
plt.legend()

plt.tight_layout()
plt.show()

# Show some example pairs
print(f"\nFirst 10 (t, F, t²) pairs:")
for i in range(10):
    print(f"t={t_stats[i]:.4f}, F={f_stats[i]:.4f}, t²={t_stats[i]**2:.4f}")
```
